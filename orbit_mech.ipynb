# Creating a collision

Using Blackswan Space's MDS software, the goal is to create a simulation which monitors the relative positions and velocities of two satellites and updates a risk assessment (SAFE $\to$ CAUTION $\to$ CRITICAL) based on the distances and closing velocities between the satellites. The risk level elevates from SAFE to CAUTION when one satellite approaches within 10 km of another, and from CAUTION to CRITICAL when the two are within 5 km of one another, or if they are within 5 km of one another and a collision is predicted to occur within 2 minutes.

When creating the satellites, the main variables that are chosen are the objects' masses, semi-major axes, eccentricities, inclinations, right ascension of the ascending node(RAAN), arguments of periapsis, true anomalies, and the central body around which they are orbiting (the Earth in this sim). We are also able to apply thrusters to the satellites, which gives control over the satellite's maximum thrust, $\Delta V$, specific impulse, burn duration limit, and fuel consumption rate. This presents a couple of options for creating a collision. We could set the two satellites on intersecting orbits by adjusting their initial positions via inclination, initial RAAN, and initial true anomaly, or we could start the satellites on a more realistic non-intersecting orbit and simulate an anomalous maneuver by applying a thrust vector to one satellite, resulting in a change in orbit which separates the violation boundary. I have chosen the latter approach.

# Initial sim conditions

The two satellites are named "Sat A" and "Sat B". Sat A will remain the stable satellite while Sat B will be forced into a collision course. The desired behaviour is to show the simulation advance normally for 2 minutes before introducing an event which changes Sat B's trajectory, causing the separation between the satellites to decrease towards the 5 km boundary. Without an evasive maneuver, the spacecraft should cross the safety boundary within the remaining simulation period.

Satellites A and B will be placed in a Low Earth Orbit (LEO) of 800 km, and will have 0 eccentricity for the sake of simplification. The initial true anomaly will then be the changed value, placing satellite B 15 km ahead of A using the relationship $$d = 2r \sin (\frac{\Delta \nu}{2})$$ where $d$ is the distance between the satellites, r is the orbital radius, and $\Delta \nu$ is the difference in true anomaly in radians. This can be rearranged then to $$ \Delta \nu = 2 arcsin (\frac{d}{2r})$$

$r$ can then be found as the radius of the earth ($R_E$) + the altitude $h$, so $r = R_E + h$.
