# Monitoring a Close-Approach Event

Using Blackswan Space's MDS software, the goal is to create a simulation which monitors the relative separation of two satellites and updates a risk assessment (SAFE $\to$ CAUTION $\to$ CRITICAL) based on the distances and closing velocities. 

The risk level is `SAFE` when the satellites are more than 10 km apart. It updates to `CAUTION` when their separation decreases to 10 km or less. The risk becomes `CRITICAL` when the two are within a 5 km safety boundary, or if their current closing velocity predicts that they will cross the 5 km boundary within the next 2 minutes.

When creating the satellites, the main orbital parameters are their masses, semi-major axes, eccentricities, inclinations, right ascension of the ascending node (RAAN), arguments of periapsis, true anomalies, and the central body around which they are orbiting. In this simulation, the central body is Earth. 

MDS also allows for thrusters to be added to the satellites. These define parameters such as maximum thrust, $\Delta V$, specific impulse, burn duration limit, and fuel consumption rate. This allows for a couple of options for creating a close-approach scenario. The satellites could be placed on intersecting orbits by adjusting their initial positions via inclination, RAAN, and true anomaly. Alternatively, we could start the satellites on a more realistic non-intersecting orbit and simulate an anomalous maneuver by applying a thrust vector to one satellite, resulting in a change in orbit which separates the violation boundary. 

I have chosen the latter approach. This provides a more controlled demonstration in which the satellites begin in a safe configuration before an anomalous manever causes their separation to decrease.

# Initial sim conditions

The two satellites are named "Sat A" and "Sat B". Sat A will remain the stable satellite while Sat B will be subjected to an anomalous maneuver. The simulation is designed to advance normally for 2 minutes before the event is introduced. The resulting trajectory change causes the separation between the satellites to decrease towards the 5 km safety boundary. 

Satellites A and B will be placed in a Low Earth Orbit (LEO) at an altitude of 800 km. An eccentricity of 0 is used to simpligy the orbital configuration. The initial true anomaly will then be the changed value, placing satellite B 15 km ahead of A using the chord-length relationship $$d = 2r \sin (\frac{\Delta \nu}{2})$$ where $d$ is the distance between the satellites, r is the orbital radius, and $\Delta \nu$ is the difference in true anomaly in radians. This can be rearranged then to 

$$\Delta \nu = 2 arcsin (\frac{d}{2r})$$

The orbital radius is calculated from the Earth's radius ($R_E$) and the orbital altitude ($h$):

$$r = R_E + h$$

The resulting angular separation is then used to set Sat B's initial true anomaly relative to Sat A.

# Creating retrograde burn

The anomalous maneuver is created by applying a retrograde thrust to Sat B. The satellite's current velocity vector is obtained from MDS and normalized to determine its direction of travel. A thrust vector pointing opposite to this velocity is then generated as $$\vec{T} = -T\frac{\vec{V}}{|\vec{v}|}$$ where $\vec{T}$ is the thrust vector, $T$ is the commanded thrust magnitude, and $\vec{v}$ is the satellite's velocity vector.

The resulting vector points opposite to Sat B's current direction of motion. Applying this thrust changes its orbital velocity and therefore causes its trajectory to deviate from the original orbit. In the simulation, this maneuver is introduced after the first 120 seconds and is applied for a limited duration.

The purpose of the maneuver is not to model a physically complete collision scenario, but to create a controlled close-approach event that allows the monitoring system to detect increasing risk and identify when an avoidance response would be required.
